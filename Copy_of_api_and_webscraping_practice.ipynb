# APIs

Application Programming Interfaces, or APIs, allow software to communicate with each other to exchange data and other functionality.

Most APIs are accessed using URL requests, much like requesting a page in a web browser. For our purposes, a URL is just a file or application that we can access using a web address (e.g., http://bc.edu). Python has features to help us make the requests.

The APIs we'll work with have URLs thst return data in raw formats such as `json` or `XML` files. These differ from the `html` files that we typically read in a web browswers. You learned about structured and unstructured data earlier. APIs return structured data; `html` files typically return unstructured data that is rendered for people to easily read.

### API example
We'll use GitHub a lot in our data science work, so let's use its API for our example. The GitHub API is accessible at `api.github.com.` This URL will return structured data files from Github.    

```python
import pandas as pd

# The `requests` library allows us to make URL requests and retrieve data from
# URLs
import requests
```

Now we can make make a request to the GitHub API to return information about a specific user. That user should be you! Enter your GitHub name in the cell below and run it to make the API request.

```python
# request data from GitHub API about your account

# FILL IN YOUR GITHUB USER NAME BELOW
github_account_name = "jilly4204" # PUT YOUR GITHUB USER NAME BETWEEN THE QUOTES"

# DON'T CHANGE THE LINE BELOW!
assert github_account_name != "", f"Make sure to add your GitHub user name to the variable githib_account_name!"

# The GitHub API URL is https://api.github.com/users/YOUR_USER_NAME.
# Put your user name in the gihub_account_name variable named
# github_account_name. We use the '+' operator below to add
# the API URL with your user name

page = requests.get('https://api.github.com/users/' + github_account_name)

# the content we get back from `requests.get` is a json file
page.content
```

JSON data is highly structured, but it's not always easy for people to read. We can make this easier by loading the data into a pandas dataframe. To do this, we call `page.json()` to return the data as a Python dictionary. We can make the dataframe from that dictionary using `from_dict`.

```python
# use the argument orient=index in the call to from_dict. Why?
# this tells pandas that the keys of the dictionary are row
# indices, and the values are the row data
df = pd.DataFrame.from_dict(page.json(), orient='index')
df
```

APIs let us collect structured data with a few Python commands. In this case, you just pulled data about your GitHub profile. # Web Scraping

<div class="alert alert-info">
Web scrapers are programs that read and extract data from websites.
</div>

The websites generally consist of text and HTML tags. Those tags define the structure and layout of the document. For instance, we see tags like `<i>` to italicize text, tags like `<table>` to define tables of data. We're going to extract a table from a Wikipedia page in this example.  

For general web scraping, the Python library called [BeautifulSoup](https://beautiful-soup-4.readthedocs.io/en/latest/) is an incredibly useful tool. But pandas has a convenient function named `read_html`. It does just that: Read HTML files so they can be worked with as dataframes.  

We'll use the Python library called `requests` again to  retrieve web site content. Let's import it and `pandas`.


```python
# requests is a library that lets us read from web pages.
import requests

# StringIO is a library that lets us treat strings as if they're files. Functions like `read_html()` want strings
# to be StringIO objects if they're not reading from actual files. You'll see an example below.
from io import StringIO

# This one should look familiar...
import pandas as pd
```

### Getting the Web Content
The first thing to do is to grab the contents of our website.   

Make a variable called `skycrapercenter_url` that holds the link to the tallest buildings page on Wikipedia. That URL is:  
[https://www.skyscrapercenter.com/buildings](https://www.skyscrapercenter.com/buildings)  

Use `requests.get()` to open the URL using `skycrapercenter_url` as its input. That will give you the webpage: Store this in a variable named `skyscraper_page`.  

```python
# site_url will hold the URL for the Wikipedia building page
skycrapercenter_url = "https://www.skyscrapercenter.com/buildings"

# get the web page using the URL request
skyscraper_page = requests.get(skycrapercenter_url)
```

`requests.get()` returns a `Response` object that has a few useful methods associated with it. `.text` is one of them.

Let's print the first 1000 characters of the skyscrapercenter.com entry. We do that by requesting the `.text` of `skyscraper_page`.

```python
# print the first 1000 characters of the web page
print(skyscraper_page.text[0:1000])
```

This is definitely HTML and not meant for human consumption. How can we find the table(s) on the page?  

`read_html` is designed to specifically look for the tables. Let's use it on the skyscraper_page text.

```python
# Find tables on the web page using read_html
# The skyscraper_page.text is wrapped in a StringIO type. This
# is an object for reading and writing files
skyscraper_tables = pd.read_html(StringIO(skyscraper_page.text))

# let's see what we get in skyscraper_tables
print("skyscraper_tables is a", type(skyscraper_tables), "of length", len(skyscraper_tables))
```

Before looking at the skyscraper_tables content, we checked its type to see that it's a list with 1 element. The length is 1 because there's only 1 table on the page. If we're lucky, that first/only item in the list is the table of skyscraper heights. Let's take a look.

```python
skyscraper_data = skyscraper_tables[0]
skyscraper_data.head()
```

### Cleaning the columns
That looks like the table we saw at [https://www.skyscrapercenter.com/buildings](https://www.skyscrapercenter.com/buildings). Except the headings are ridiculously long. `read_html` extracted the header names **and** additional infomation text buried in the headers (you'll see the text by rolling over the ⓘ symbol in the header row). We just want the column titles, so let's clean up the dataframe.

```python
# make a list with new column names
columns = ['rank', 'name', 'city', 'status', 'completion', 'height', 'floors', 'material', 'function']

skyscraper_data = skyscraper_tables[0]

# set the columns to the columns list
skyscraper_data.columns = columns

# see if it worked
skyscraper_data.head()
```

And there you have it, a Pandas dataframe with all of the data from the skyscraper webpage! That was a lot of work, so let's save the dataframe so we don't need to scrape the web page again. We use `to_csv` to convert the dataframe into a CSV file. We also add the argument `index=False`; this will drop the index values (0,1,2...) from the CSV file. They'll return when we reload the file into a dataframe.

```python
skyscraper_data.to_csv("tallest_skyscrapers.csv", index=False)
```
